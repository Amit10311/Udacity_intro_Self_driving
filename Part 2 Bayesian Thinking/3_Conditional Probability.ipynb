# **3.2 Learning Objectives - Conditional Probability**

We use the notation where

* $P(A)$ means "the probability of A"
* $P(¬A)$ means "the probability of NOT A"
* $P(A,B)$ means "the probability of A **and** B"
* $P(A|B)$ means "the probability of A **given** B

``
* $P(A|B)= \frac{P(A\cap	B)}{P(B)}$

* $P(A\cap B)= A-B $



### **1. Conditional Probbailty Rule**
  > **$P(A\cap B)= P(A)*P(B|A)= P(B)* P(A/B) $**

 **Ans:** $P(A,B)= P(A)*P(B)= 0.2*0.1 =0.02$

## **2. Rule of Total Probabilty**

> $P(E) = P(A \cap E) + P( B\cap E)$
 >> $ = P(A)* P(E|A) + P(B) * P(E|B) $

## `QUESTION 4 OF 5`

**Ques:** You grab one of these two coins at random (equally likely that you grabbed coin 1 or 2) and you flip it. What's the probability it comes up heads?

**Ans:**  $P(E) = P(H1)* P(E|H1) + P(H2) * P(E|H2) $

 >> $ =  0.5*0.5+ 0.9*0.5 = 0.7 $

## `QUESTION 5 OF 5`

**Ques:** You grab a coin at random and flip it twice.

What's the probability that it comes up tails both times?

**Ans:**  $P(E) = P(T1,T1)* P(E|T1) + P(T2,T2) * P(E|T2) $

 >> $ =  0.5*0.5*0.5+ 0.1*0.1*0.5 = 0.13 $

### Law of Total Probability

A useful law when dealing with conditional probability is the Law of Total Probability.

If $B_1, B_2, B_3, ...$ ,... is a partition of a sample space $S$, then for any event $A$:

$P(A)$ =$\sum_i P(A \cap B_i)$ =$ \sum_i P(A | B_i) * P(B_i)P(A)$
